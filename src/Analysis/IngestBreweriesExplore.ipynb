```python
# Pre-defined schema usually provided in cell 1
from pyspark.sql.types import StructType, StructField, StringType

schema = StructType([
    StructField("question", StringType(), True),
    StructField("answer", StringType(), True)
])

# Initialize empty answers_df
answers_df = spark.createDataFrame([], schema)
```

```python
import pyspark.sql.functions as F

# Load your ingested dataset
abfss_path = "abfss://rawdata@manuelstorage01.dfs.core.windows.net/Bronze/Breweries"
df = spark.read.option("header", "true").csv(abfss_path)
```

```python
q1_val = str(df.count())

new_row = spark.createDataFrame([("How many rows are in your ingested dataset?", q1_val)], schema)
answers_df = answers_df.union(new_row)
```

```python
q2_val = str(len(df.columns))

new_row = spark.createDataFrame([("How many columns are in your ingested dataset??", q2_val)], schema)
answers_df = answers_df.union(new_row)
```

```python
q3_val = str(df.select("country").distinct().count())

new_row = spark.createDataFrame([("How many countries are in your dataset??", q3_val)], schema)
answers_df = answers_df.union(new_row)
```

```python
q4_val = str(df.filter(F.lower(F.col("brewery_type")) == "micro").count())

new_row = spark.createDataFrame([("How many micro breweries are in your dataset??", q4_val)], schema)
answers_df = answers_df.union(new_row)
```
